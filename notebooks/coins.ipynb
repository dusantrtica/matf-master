```python
from src.algo.data import *
from ortools.sat.python import cp_model
```

```python
model = cp_model.CpModel()
solver = cp_model.CpSolver()
```

```python
while amount := int(input("Enter number: ")):
    if amount == 0:
        break

    x1 = model.NewIntVar(0, 99, "1")
    x2 = model.NewIntVar(0,49, "2")
    x5 = model.NewIntVar(0, 19, "5")
    x10 = model.NewIntVar(0, 9, "10")
    x20 = model.NewIntVar(0, 4, "20")

    model.Add(x1*1 + x2*2 + x5*5 + x10*10 + x20*20 == amount)
    model.Minimize(x1 + x2 + x5 + x10 + x20)

    print(solver.Solve(model))

    print(solver.Value(x1))
    print(solver.Value(x2))
    print(solver.Value(x5))
    print(solver.Value(x10))
    print(solver.Value(x20))

    
```
