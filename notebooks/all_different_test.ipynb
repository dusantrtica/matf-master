```python
from ortools.sat.python import cp_model
```

```python
model = cp_model.CpModel()
solver = cp_model.CpSolver()
```

```python
n = 10
variables = [model.NewIntVar(lb=0, ub=100, name=f"variable_{i}") for i in range(n)]
```

```python
model.AddAllDifferent(variables)
```

```python
solver.solve(model)
```

```python
for i in range(n):
    print(solver.Value(variables[i]))
```
